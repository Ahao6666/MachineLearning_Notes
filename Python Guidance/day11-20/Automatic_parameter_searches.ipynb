# 自动参数搜索示例

这个notebook演示了如何使用RandomizedSearchCV进行随机参数搜索，以优化随机森林回归模型的超参数。

主要内容包括：
1. 生成合成回归数据集
2. 划分训练集和测试集
3. 定义参数搜索空间
4. 使用RandomizedSearchCV进行参数搜索
5. 查看最佳参数并评估模型性能

## 导入所需库

我们需要导入以下模块：
- `make_regression`: 生成合成回归数据集
- `RandomForestRegressor`: 随机森林回归模型
- `RandomizedSearchCV`: 随机参数搜索交叉验证
- `train_test_split`: 划分训练集和测试集
- `randint`: 用于生成随机整数参数分布

```python
# 生成合成数据集的工具
from sklearn.datasets import make_regression
# 随机森林回归模型
from sklearn.ensemble import RandomForestRegressor
# 随机参数搜索交叉验证工具
from sklearn.model_selection import RandomizedSearchCV
# 数据集划分工具
from sklearn.model_selection import train_test_split
# 用于生成随机整数分布
from scipy.stats import randint
```

```python
# 创建合成回归数据集
X, y = make_regression(n_samples=20640,      # 样本数量
                       n_features=8,         # 特征数量
                       noise=0.1,            # 噪声水平，增加数据难度
                       random_state=0)       # 随机种子，保证结果可复现
print(f"X shape: ",X.shape)                 # 输出特征矩阵形状
print(f"y shape: ",y.shape)                 # 输出目标向量形状
```

## 生成合成数据集

使用`make_regression`生成一个包含20640个样本，8个特征的合成回归数据集，并添加一定的噪声。

```python
# 将数据集按3:1比例划分为训练集和测试集
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)
print(f"X_train shape: ",X_train.shape)     # 训练集特征形状
print(f"X_test shape: ",X_test.shape)       # 测试集特征形状
print(f"y_train shape: ",y_train.shape)     # 训练集目标形状
print(f"y_test shape: ",y_test.shape)       # 测试集目标形状
```

## 划分训练集和测试集

将生成的数据集划分为训练集和测试集，用于模型训练和评估。

```python
# 定义需要搜索的参数空间
param_distributions = {
    'n_estimators': randint(1, 5),  # 树的数量，随机取值1-4
    'max_depth': randint(5, 10)     # 树的最大深度，随机取值5-9
}

# 创建随机搜索交叉验证对象
search = RandomizedSearchCV(
    estimator=RandomForestRegressor(random_state=1),  # 基础模型
    n_iter=5,                                         # 尝试5组不同的参数组合
    param_distributions=param_distributions,          # 参数搜索空间
    random_state=0                                    # 随机种子，保证结果可复现
)

# 在训练集上进行参数搜索
search.fit(X_train, y_train)

# 输出搜索得到的最佳参数组合
print("最佳参数：", search.best_params_)

# 使用最佳参数的模型在测试集上评估性能
score = search.score(X_test, y_test)
print("测试集R²得分：", score)
```

## 参数搜索与模型训练

### 定义参数搜索空间
我们定义了两个超参数的搜索范围：
- `n_estimators`: 随机森林中树的数量，范围从1到4
- `max_depth`: 树的最大深度，范围从5到9

### 创建搜索对象并训练
使用`RandomizedSearchCV`进行随机参数搜索，总共尝试5种不同的参数组合。

### 查看最佳参数
搜索完成后，可以得到最佳的参数组合。

### 模型评估
使用搜索得到的最佳参数模型在测试集上评估性能。
